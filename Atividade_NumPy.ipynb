# Atividade Prática NumPy — Varejo

## Nível 1: Faturamento Semanal e Filtros

```python
import numpy as np

pecas_vendidas = np.array([150, 120, 90, 210, 300, 250, 180])
pecas_vendidas
```

```python
faturamento = pecas_vendidas * 50
faturamento
```

```python
dias_pico = pecas_vendidas[pecas_vendidas > 200]
dias_pico
```

## Nível 2: Múltiplas Filiais e Falhas de Sistema

```python
vendas = np.array([
    [200, 220, np.nan, 250],
    [150, 180, 160, 190],
    [300, 310, 290, 330]
])
vendas
```

```python
total_por_loja = vendas.sum(axis=1)
total_por_loja
```

```python
media_geral = np.nanmean(vendas)
media_geral
```

```python
meta = np.where(vendas >= 200, 'Meta Atingida', 'Abaixo')
meta
```

## Nível 3: Categorização e Marketing

```python
rng = np.random.default_rng(42)
amostra = rng.choice(['Eletrônicos', 'Roupas', 'Casa'], size=50)
amostra
```

```python
valores, contagens = np.unique(amostra, return_counts=True)
valores, contagens
```

```python
campanhas = np.array([12000, 45000, 23000, 89000, 31000])
campanha_campea = np.argmax(campanhas)
campanha_campea
```

# Atividade Prática NumPy — Organização e Transformação de Dados

## Nível 1: Sequências e Tipagem

```python
dias_mes = np.arange(1, 31)
dias_mes
```

```python
metas = np.linspace(20000, 30000, 5)
metas
```

```python
estoque = np.array([10.5, 20.1, 30.9]).astype(int)
estoque
```

```python
ultimos_5 = dias_mes[-5:][::-1]
ultimos_5
```

## Nível 2: Redimensionamento e Proteção de Dados

```python
visitas = np.arange(1, 13)
trimestres = visitas.reshape(4, 3)
trimestres
```

```python
primeiro_sem = np.arange(1, 7).reshape(1, 6)
segundo_sem = np.arange(7, 13).reshape(1, 6)
ano_completo = np.vstack((primeiro_sem, segundo_sem))
ano_completo
```

```python
projecao = trimestres[0:1].copy()
projecao
```

```python
achatado = trimestres.ravel()
achatado
```

## Nível 3: Ranking, Broadcasting e Sistemas Lineares

```python
notas = np.array([85, 92, 78, 95, 88])
ranking = np.argsort(notas)
ranking
```

```python
precos = np.array([100, 200, 300])
precos_coluna = precos[:, np.newaxis]
precos_coluna
```

```python
A = np.array([[2, 1], [1, -1]])
b = np.array([500, 100])
solucao = np.linalg.solve(A, b)
solucao
```
